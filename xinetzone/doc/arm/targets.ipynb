# TVM ARM 目标平台测试文档

```python
import tvm
from tvm.script import ir_module, relax as R

@ir_module
class ReshapeModule:
    @R.function
    def main(x: R.Tensor((1, 80, 800), "float32")) -> R.Tensor((1, 5, 800, 1, 16), "float32"):
        gv0 = R.reshape(x, (1, 5, 16, 800))
        gv1 = R.permute_dims(gv0, axes=(0, 1, 3, 2))
        gv2 = R.expand_dims(gv1, axis=3)
        return gv2

# 编译 Relax 模块
target = tvm.target.Target("llvm")
ex = tvm.compile(ReshapeModule, target=target)

# 导出为共享库
ex.export_library("reshape_scenario_a.so")
```
